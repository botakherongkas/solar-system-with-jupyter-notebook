```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib as mpl

# Menghindari batas ukuran animasi HTML
mpl.rcParams["animation.embed_limit"] = 25

# -----------------------------
# Parameter simulasi
# -----------------------------
TOTAL_TAHUN = 230_000_000
JUMLAH_FRAME = 180

waktu = np.linspace(0, TOTAL_TAHUN, JUMLAH_FRAME)

# -----------------------------
# Data planet
# -----------------------------
nama_planet = [
    "Matahari",
    "Merkurius",
    "Venus",
    "Bumi",
    "Mars",
    "Jupiter",
    "Saturnus",
    "Uranus",
    "Neptunus"
]

jarak_orbit = np.array([
    0,
    2,
    4,
    6,
    8,
    12,
    16,
    20,
    24
], dtype=float)

warna = [
    "gold",
    "dimgray",
    "orange",
    "royalblue",
    "firebrick",
    "peru",
    "khaki",
    "lightblue",
    "darkblue"
]

ukuran = [
    16,
    4,
    5,
    5,
    4,
    8,
    7,
    7,
    7
]

# Banyak putaran selama animasi
putaran = [
    0,
    24,
    18,
    12,
    8,
    4,
    3,
    2,
    1
]

# -----------------------------
# Menghitung posisi planet
# -----------------------------
x = np.zeros((len(nama_planet), JUMLAH_FRAME))
y = np.zeros((len(nama_planet), JUMLAH_FRAME))

for i in range(1, len(nama_planet)):

    sudut = 2 * np.pi * putaran[i] * waktu / TOTAL_TAHUN

    x[i] = jarak_orbit[i] * np.cos(sudut)
    y[i] = jarak_orbit[i] * np.sin(sudut)

# -----------------------------
# Membuat gambar
# -----------------------------
fig, ax = plt.subplots(figsize=(9, 9))

fig.patch.set_facecolor("black")
ax.set_facecolor("black")

ax.set_xlim(-28, 28)
ax.set_ylim(-28, 28)
ax.set_aspect("equal")

ax.set_xticks([])
ax.set_yticks([])

# Orbit planet
for r in jarak_orbit[1:]:

    orbit = plt.Circle(
        (0, 0),
        r,
        fill=False,
        color="white",
        linestyle="--",
        alpha=0.25
    )

    ax.add_artist(orbit)

# Planet
planet = []

for i in range(len(nama_planet)):

    titik, = ax.plot(
        [],
        [],
        "o",
        color=warna[i],
        markersize=ukuran[i]
    )

    planet.append(titik)

# Label
label = []

for i in range(len(nama_planet)):

    teks = ax.text(
        0,
        0,
        nama_planet[i],
        color="white",
        fontsize=8
    )

    label.append(teks)

# Jejak orbit
jejak = []

for i in range(len(nama_planet)):

    garis, = ax.plot(
        [],
        [],
        color=warna[i],
        linewidth=1,
        alpha=0.45
    )

    jejak.append(garis)

# -----------------------------
# Fungsi animasi
# -----------------------------
def update(frame):

    for i in range(len(nama_planet)):

        planet[i].set_data(
            [x[i, frame]],
            [y[i, frame]]
        )

        jejak[i].set_data(
            x[i, :frame+1],
            y[i, :frame+1]
        )

        label[i].set_position(
            (
                x[i, frame] + 0.4,
                y[i, frame] + 0.4
            )
        )

    ax.set_title(
        f"Simulasi Tata Surya\n"
        f"Waktu: {int(waktu[frame]):,} tahun",
        color="white",
        fontsize=14
    )

    return planet + jejak + label

# -----------------------------
# Membuat animasi
# -----------------------------
animasi = FuncAnimation(
    fig,
    update,
    frames=JUMLAH_FRAME,
    interval=70,
    blit=True
)

plt.close(fig)

HTML(animasi.to_jshtml())
```
