```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib as mpl
mpl.rcParams['animation.embed_limit']=25
TOTAL_TAHUN=230_000_000
JUMLAH_FRAME=180
waktu=np.linspace(0,TOTAL_TAHUN,JUMLAH_FRAME)
nama=['Matahari','Merkurius','Venus','Bumi','Mars','Jupiter','Saturnus','Uranus','Neptunus']
jarak=np.array([0,2,4,6,8,12,16,20,24],float)
warna=['gold','dimgray','orange','royalblue','firebrick','peru','khaki','lightblue','darkblue']
ukuran=[16,4,5,5,4,8,7,7,7]
putaran=[0,24,18,12,8,4,3,2,1]
x=np.zeros((len(nama),JUMLAH_FRAME)); y=np.zeros_like(x)
for i in range(1,len(nama)):
    sudut=2*np.pi*putaran[i]*waktu/TOTAL_TAHUN
    x[i]=jarak[i]*np.cos(sudut)
    y[i]=jarak[i]*np.sin(sudut)
fig,ax=plt.subplots(figsize=(9,9))
fig.patch.set_facecolor('black'); ax.set_facecolor('black')
ax.set_xlim(-28,28); ax.set_ylim(-28,28); ax.set_aspect('equal')
ax.set_xticks([]); ax.set_yticks([])
for r in jarak[1:]: ax.add_artist(plt.Circle((0,0),r,fill=False,color='white',linestyle='--',alpha=0.25))
planet=[]; label=[]; jejak=[]
for i in range(len(nama)):
    p,=ax.plot([],[],'o',color=warna[i],markersize=ukuran[i]); planet.append(p)
    label.append(ax.text(0,0,nama[i],color='white',fontsize=8))
    g,=ax.plot([],[],color=warna[i],linewidth=1,alpha=0.45); jejak.append(g)
def update(frame):
    for i in range(len(nama)):
        planet[i].set_data([x[i,frame]],[y[i,frame]])
        jejak[i].set_data(x[i,:frame+1],y[i,:frame+1])
        label[i].set_position((x[i,frame]+0.4,y[i,frame]+0.4))
    ax.set_title(f'Simulasi Tata Surya\nWaktu: {int(waktu[frame]):,} tahun',color='white')
    return planet+label+jejak
animasi=FuncAnimation(fig,update,frames=JUMLAH_FRAME,interval=70,blit=True)
plt.close(fig)
HTML(animasi.to_jshtml())
```
