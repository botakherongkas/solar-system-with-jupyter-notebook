# Simple Solar System Simulation
A Jupyter Notebook that visualizes a simplified Solar System simulation using Python.


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib as mpl

mpl.rcParams['animation.embed_limit'] = 25

TOTAL_YEARS = 230_000_000
NUM_FRAMES = 180
time = np.linspace(0, TOTAL_YEARS, NUM_FRAMES)

planet_names = [
    'Sun', 'Mercury', 'Venus', 'Earth', 'Mars',
    'Jupiter', 'Saturn', 'Uranus', 'Neptune'
]

orbit_radius = np.array([0, 2, 4, 6, 8, 12, 16, 20, 24], dtype=float)

planet_colors = [
    'gold', 'dimgray', 'orange', 'royalblue',
    'firebrick', 'peru', 'khaki', 'lightblue', 'darkblue'
]

planet_sizes = [16, 4, 5, 5, 4, 8, 7, 7, 7]

orbital_speed = [0, 24, 18, 12, 8, 4, 3, 2, 1]

x = np.zeros((len(planet_names), NUM_FRAMES))
y = np.zeros_like(x)

for i in range(1, len(planet_names)):
    angle = 2 * np.pi * orbital_speed[i] * time / TOTAL_YEARS
    x[i] = orbit_radius[i] * np.cos(angle)
    y[i] = orbit_radius[i] * np.sin(angle)

fig, ax = plt.subplots(figsize=(9, 9))
fig.patch.set_facecolor('black')
ax.set_facecolor('black')

ax.set_xlim(-28, 28)
ax.set_ylim(-28, 28)
ax.set_aspect('equal')
ax.set_xticks([])
ax.set_yticks([])

for radius in orbit_radius[1:]:
    ax.add_artist(
        plt.Circle(
            (0, 0),
            radius,
            fill=False,
            color='white',
            linestyle='--',
            alpha=0.25
        )
    )

planets = []
labels = []
trails = []

for i in range(len(planet_names)):
    planet, = ax.plot([], [], 'o', color=planet_colors[i], markersize=planet_sizes[i])
    planets.append(planet)

    labels.append(
        ax.text(0, 0, planet_names[i], color='white', fontsize=8)
    )

    trail, = ax.plot([], [], color=planet_colors[i], linewidth=1, alpha=0.45)
    trails.append(trail)

def update(frame):
    for i in range(len(planet_names)):
        planets[i].set_data([x[i, frame]], [y[i, frame]])
        trails[i].set_data(x[i, :frame+1], y[i, :frame+1])
        labels[i].set_position((x[i, frame] + 0.4, y[i, frame] + 0.4))

    ax.set_title(
        f'Solar System Simulation\nSimulation Time: {int(time[frame]):,} years',
        color='white'
    )

    return planets + labels + trails

animation = FuncAnimation(
    fig,
    update,
    frames=NUM_FRAMES,
    interval=70,
    blit=True
)

plt.close(fig)
HTML(animation.to_jshtml())
```
